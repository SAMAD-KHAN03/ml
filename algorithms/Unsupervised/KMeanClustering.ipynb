```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# -------------------------------------------------------------
# # Bollywood Songs Dataset
# -------------------------------------------------------------

songs = {
    "Song": [
        "Tum Hi Ho", "Channa Mereya", "Kesariya", "Shayad",
        "Raataan Lambiyan", "Apna Bana Le", "Tera Ban Jaunga",
        "Kal Ho Naa Ho", "Agar Tum Saath Ho", "Pee Loon",

        "Kala Chashma", "Malhari", "Zingaat", "Kar Gayi Chull",
        "Badtameez Dil", "Nashe Si Chadh Gayi", "Ghungroo",
        "Jhoome Jo Pathaan", "Dilbar", "Param Sundari",

        "Senorita", "Gallan Goodiyan", "London Thumakda",
        "Morni Banke", "Bom Diggy Diggy", "Saturday Saturday",

        "Bekhayali", "Kabira", "Phir Le Aya Dil",
        "Ae Dil Hai Mushkil", "Tujhe Kitna Chahne Lage",

        "Subah Hone Na De", "Abhi Toh Party Shuru Hui Hai",
        "The Breakup Song", "Cutiepie", "Hookah Bar",

        "Tum Se Hi", "Pehla Nasha", "Janam Janam",
        "Gerua", "Ve Maahi", "Ranjha"
    ],
    # Adding necessary placeholder features (e.g., Tempo, Energy, Danceability)
    # to match the K-Means + PCA pipeline shown in your output console.
    "Tempo": [
        70, 75, 80, 72, 78, 74, 76, 82, 73, 77,        # Romantic / Melancholy
        130, 135, 140, 122, 126, 118, 120, 116, 124, 128, # High Energy / Dance
        125, 128, 132, 122, 126, 130,                     # Wedding / Party
        85, 80, 75, 82, 78,                               # Intense / Emotional
        128, 130, 124, 126, 128,                          # Club / Party Upbeat
        90, 85, 88, 84, 82, 80                            # Melodic / Soft Pop
    ],
    "Energy": [
        0.4, 0.45, 0.5, 0.42, 0.48, 0.46, 0.47, 0.52, 0.41, 0.49,
        0.9, 0.95, 0.98, 0.88, 0.92, 0.85, 0.87, 0.84, 0.89, 0.91,
        0.8, 0.85, 0.88, 0.82, 0.86, 0.87,
        0.7, 0.5, 0.4, 0.65, 0.55,
        0.88, 0.92, 0.84, 0.86, 0.90,
        0.6, 0.55, 0.62, 0.58, 0.52, 0.50
    ]
}

# Create DataFrame
df = pd.DataFrame(songs)

# -------------------------------------------------------------
# # Clustering and PCA Pipeline
# -------------------------------------------------------------

# 1. Feature Selection and Scaling
features = df[["Tempo", "Energy"]]
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)

# 2. Apply K-Means Clustering (e.g., 4 clusters based on your plot's colors)
kmeans = KMeans(n_clusters=4, random_state=42)
df["Cluster"] = kmeans.fit_predict(scaled_features)

# Print a subset of songs matching your console output example
print("Console Output Example (Songs in a cluster):")
print(df[df["Cluster"] == 0]["Song"].to_string(index=False))

# 3. PCA Dimensionality Reduction for 2D Visualization
pca = PCA(n_components=2)
pca_features = pca.fit_transform(scaled_features)
df["PCA1"] = pca_features[:, 0]
df["PCA2"] = pca_features[:, 1]

# 4. Plotting the results
plt.figure(figsize=(10, 6))
scatter = plt.scatter(df["PCA1"], df["PCA2"], c=df["Cluster"], cmap="viridis", edgecolors="k")

# Plot Cluster Centroids
centroids = pca.transform(kmeans.cluster_centers_)
plt.scatter(centroids[:, 0], centroids[:, 1], marker="X", s=150, c="blue", label="Centroids")

plt.title("Bollywood Songs Clustering using PCA + K-Means")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```
